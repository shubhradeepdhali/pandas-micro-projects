### Create a new column score_level
"High" if total_score ≥ 90 , "Medium" if 70 ≤ total_score < 90, "Low" if total_score < 70

```python
def level (x):
    if x>= 90:
        return "High"
    elif x>= 70:
        return "Medium"
    elif x< 70:
        return "Low"
```

```python
df['score_level'] = df['total_score'].apply(level)
df
```

### Which student has the highest total_score?

### Create a new column study_category
"Light" if weekly_self_study_hours < 10,
"Moderate" if 10 ≤ hours ≤ 20,
"Heavy" if hours > 20

```python
def category(study_hours):
    if study_hours<10:
        return 'Light'
    elif study_hours <20:
        return 'Medium'
    else:
        return 'Heavy'
```

```python
df['Category'] = df['weekly_self_study_hours'].apply(category)
df
```
