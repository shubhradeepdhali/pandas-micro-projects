```python
# load dataset and mention encoding
df = pd.read_csv('data.csv',encoding = 'unicode_escape')
df
```

```python
# handling missing values
df.isna().sum()
```

```python
# drop missing values
df = df.dropna(subset = ["CustomerID","Description"],how = "any") # put how = 'any' for INTERSECTION and 'all' for UNION dropping
df
```

```python
# 541909 - 406829 = 135080
print(f"Rows removed: {541909 - 406829}")
```

```python
# remove negative prices and quantities
df = df[(df["Quantity"]>0)&(df["UnitPrice"])]
df
```

```python
# 406829 - 397884 = 8945 rows dropped
df.info()
```

```python
df["CustomerID"] = df['CustomerID'].astype(int)
df
# here we convert CustomerID data type from float to int
```

```python
# pandas date time conversion
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df.info()
```

```python
# adding new column for finding TotalRevenue
df['TotalRevenue'] = df['Quantity'] * df['UnitPrice']
df.head()
```

```python
df = df.set_index('InvoiceNo')
df.head()
```

```python
# new cleaned dataset
df.info()
```

```python
# Keeps 'InvoiceNo' as the first column in the file
df.to_csv('data_cleaned.csv')
```
